```python
from trimesh import load, viewer
import numpy as np # linear algebra
import pandas as pd # data processing, CSV file I/O (e.g. pd.read_csv)
from datetime import datetime


#Path to project files
pathf = r'C:\\Users\\Downloads\check\\'
v1name = 'Projekt Version1_25052022'
v2name = 'Projekt_Version2_17052022'
project_v1 = pd.read_csv(pathf+v1name+'.csv', low_memory=False, index_col = 0)
project_v2 = pd.read_csv(pathf+v2name+'.csv', low_memory=False, index_col = 0)


df1 = project_v1.compare(project_v2, align_axis=1).rename(columns={'self': 'Version1', 'other': 'Version2'}, level=-1)
df1['Volume'] = df1['Volume'].replace('m³', '', regex=True)
df1['Area'] = df1['Area'].replace('m²', '', regex=True)

df1[('Volume','Version1')] = pd.to_numeric(df1[('Volume','Version1')])
df1[('Volume','Version2')] = pd.to_numeric(df1[('Volume','Version2')])
df1[('Area','Version1')] = pd.to_numeric(df1[('Area','Version1')])
df1[('Area','Version2')] = pd.to_numeric(df1[('Area','Version2')])

df2 = pd.DataFrame(df1, columns=[('Volume','Version1'), ('Volume','Version2')])


# Form geometries for individual groups - in this case, groups of modified elements

group_ids = df2.index
#print(group_ids)

group_ids_str = []
for el in group_ids:
    group_ids_str.append(str(el))
#group_ids_str

import os
import xml.etree.ElementTree as ET
import copy

path = r'C:\\Users\\Downloads\check\\'
file = 'Projekt Version1_25052022.dae'
fileObject = open(path+'/'+file, "r")
print(fileObject)

transparency_ratio = '0.1'
color_for_differences =  '0.8 0.3 0.1 1'


ET.register_namespace("", "http://www.collada.org/2005/11/COLLADASchema")

tree = ET.parse(fileObject)
root = tree.getroot()
#removeList = [instance_geometry]
geom_list = []




#material_list
material_list = []
effect_list = []
for node in root.findall('.//{http://www.collada.org/2005/11/COLLADASchema}node'):
    #print(a.attrib)
    
    if node.attrib['id'] in group_ids_str:
        url = list(node)[0].get('url')
        geom_list.append(url[1:])

        #add material id to material list
        target = list(node)[0][0][0][0].attrib['target']
        list(node)[0][0][0][0].attrib['target'] = target.replace('#ID', '#GREEN')
  
        if not target[1:] in material_list:
        	material_list.append(target[1:])
        	effect_list.append(target[1:].replace('material', 'effect'))
        

library_materials = root.find(".//{http://www.collada.org/2005/11/COLLADASchema}library_materials")

for material in material_list:

	el = ET.Element('material')
	value = material.replace('ID', 'GREEN')
	el.set('id', value)
	el.set('name', value)
	s_el = ET.SubElement(el, 'instance_effect')
	s_el.set('url', '#' + value.replace('material', 'effect'))
	library_materials.insert(0, el)


library_effects = root.find(".//{http://www.collada.org/2005/11/COLLADASchema}library_effects")	
for effect in root.findall(".//{http://www.collada.org/2005/11/COLLADASchema}effect"):

	id_ = effect.attrib['id']
	if id_ in effect_list:

		green_effect = copy.deepcopy(effect)
		green_effect.attrib['id'] = id_.replace('ID', 'GREEN')

		ambient_color = green_effect.find('.//{http://www.collada.org/2005/11/COLLADASchema}ambient/{http://www.collada.org/2005/11/COLLADASchema}color')
		ambient_color.text = color_for_differences

		diffuse_color = green_effect.find('.//{http://www.collada.org/2005/11/COLLADASchema}diffuse/{http://www.collada.org/2005/11/COLLADASchema}color')
		diffuse_color.text = color_for_differences

		library_effects.insert(0, green_effect)

	transparency = ET.Element('transparency')
	float_ = ET.SubElement(transparency, 'float')
	float_.text = transparency_ratio
	effect[0][0][0].insert(0, transparency)

file_name = 'C:\\Users\\Artem\\Downloads\\check\\Checking_Changes_Version2.dae'
tree.write(file_name)
#path = r"C:\BIMJSON\Red Sea\Reports\walls.dae"
scene_with_mesh = load(file_name, force='scene')
viewer.notebook.scene_to_notebook(scene_with_mesh, height=800)
```
